**Data Visualization with Pandas: Example-1: Line Plot - Barchat - Histogram - Scatter - Boxplot:**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Data
ww2data = {
    'Battle': [
        'Battle of Stalingrad',
        'Battle of Midway',
        'Battle of El Alamein',
        'Battle of Kursk',
        'D-Day (Normandy)',
        'Battle of Berlin',
        'Battle of Britain'
    ],

    'Year': [
        1942,
        1942,
        1942,
        1943,
        1944,
        1945,
        1940
    ],

    'Casualties': [          # Combined causalities (approx)
        2000000,   # Stalingrad
        3070,      # Midway
        35000,     # El Alamein
        800000,    # Kursk
        425000,    # D-Day (full Normandy campaign)
        1200000,   # Berlin
        90000      # Battle of Britain
    ],

    'Outcome': [
        'Soviet Victory',
        'Allied Victory',
        'Allied Victory',
        'Soviet Victory',
        'Allied Victory',
        'Soviet Victory',
        'Allied Victory'
    ]
}

df = pd.DataFrame(ww2data)
df = df.sort_values('Year')

print(df.head())

# Line plot
df.plot(x='Year', y='Casualties', kind='line', marker = 'o', title='WW2 Casualties Over Years')
plt.show()

# Bar chart

df.plot(x='Battle', y='Casualties', kind='bar', title='WW2 Casualties by Battle')
plt.xticks(rotation=45, ha='right')
plt.ylabel("Casualties")
plt.tight_layout()
plt.show()

# Histogram

df['Casualties'].plot(kind='hist', bins=5, title='Distribution of WW2 Casualties', color= "maroon")
plt.show()

# Scatter plot

plotting = df.plot(x='Year', y='Casualties', kind='scatter', title='Relationship between Year and Casualties')
for i, row in df.iterrows():
    plotting.text(row['Year'], row['Casualties'], row['Battle'], fontsize=8)

plt.show()

# Box plot

df.boxplot(column='Casualties')
plt.title('Distribution of WW2 Casualties')
plt.show()

```

**Calculate the percentage of the total fare revenue came from each passenger class.**

1: Calculating the total Fare paid across all the passengers in the titanic dataset.

```python
total_fare_paid = titanic_df['Fare'].sum()
print(f"Total Fare paid by all the passengers: £ {total_fare_paid}")
```

2: Subsetting for passengers in first class and calculate their total fare.

```python
# first_class_fare = titanic_df.loc[titanic_df['Pclass'] == 1]['Fare'].sum()

# From group-by approach
fare_by_class = titanic_df.groupby('Pclass')['Fare'].sum()

first_class_fare = fare_by_class[1]
print(f"Total Fare paid by first class passengers: £ {first_class_fare}")
```

3: Subsetting same for second class and third class

```python
# second_class_fare = titanic_df.loc[titanic_df['Pclass'] == 2]['Fare'].sum()

# From group-by approach
second_class_fare = fare_by_class[2]
print(f"Total Fare paid by second class: £ {second_class_fare}")

# third_class_fare = titanic_df.loc[titanic_df['Pclass'] == 3]['Fare'].sum()

# From group-by approach
third_class_fare = fare_by_class[3]
print(f"Total Fare paid by third class: £ {third_class_fare}")
```

4: Combining all the fare totals into a list

```python
total_fare_list = [first_class_fare, second_class_fare, third_class_fare]
print(f"Total Fare paid by each class: \n{total_fare_list}")
```

5: Calculating the percentage of the total fare revenue came frome each passenger class.

```python
fare_proportion = [each_fare / total_fare_paid for each_fare in total_fare_list]

for i, p in enumerate(fare_proportion):
    print(f"Percentage of total fare paid by class {i+1}: {p* 100:.2f}%")
```

Extra: Visual representation

```python
import seaborn as sns
import matplotlib.pyplot as plt

fare_df = pd.DataFrame({
    'Pclass': ['First Class', 'Second Class', 'Third Class'],
    'Proportion': [p * 100 for p in fare_proportion]
})

plt.figure(figsize=(8, 6))
sns.barplot(data = fare_df, x = 'Pclass', y= 'Proportion' )
plt.title('Proportion of Total Fare Paid by Passenger Class')
plt.ylabel('Proportion (%)')
plt.xlabel('Passenger Class')

for index, row in fare_df.iterrows():
    plt.text(index, row['Proportion'], f"{row['Proportion']:.2f}%",
             ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

**Question: 2- What percent of the total number of passengers on the Titanic belonged to each age group(child, adult, senior)**

1: Creating a new column, age_group that categorizes passengers into "child", "adult" and "senior"

```python
# creating a list for age range from 0-17, 18-64, 65+
age_range = [0, 17, 64, 150]
age_group = ["child", "adult", "senior"]
titanic_df['age_group'] = pd.cut(titanic_df['Age'], bins=age_range, labels=age_group)
titanic_df.head()
```

2: Calculate the total number of passengers on the titanic

```python
total_passenger = len(titanic_df)
print(f"Total number of passengers: {total_passenger}")
```

3: Count the number of passengers on the titanic.

```python
age_group_count = titanic_df.groupby('age_group')['age_group'].count()
print(f"Number of passengers in each age group: \n{age_group_count}")
```

4: Dividing the count of each age group by the total number of passengers to get the proportion of passengers in each age group.

```python
age_group_proportion = (age_group_count / total_passenger)
print("Proportion of passengers in each age group:")
print(age_group_proportion)
```

5: Displaying the proportion as a percentage.

```python
print(f"Percentage of passengers in each age group: \n{age_group_proportion *100}")
```
