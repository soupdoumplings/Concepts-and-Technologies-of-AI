Example 2: Data Structure - Index: Types of Index

```python
import pandas as pd
series = pd.Series([10, 20, 30, 40])

# printing default index of the series
print(series.index)

print('_'*50)
# for printing user defined index
series = pd.Series([10, 20, 30, 40], index = ['i', 'ii', 'iii', 'iv'])
print(series)

print('_'*50)
# for printing datetime index
dates = pd.date_range('2025-11-24', periods = 4)
series = pd.Series([10, 20, 30, 40], index = dates)
print(series)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
import numpy as np
iris = load_iris()
iris_df = pd.DataFrame(data = np.c_[iris['data'], iris['target']], columns = iris['feature_names'] + ['target'])
np.random.seed(42)
mask = np.random.rand(*iris_df.shape) < 0.1
iris_df[mask] = np.nan
print("Missing values in Iris Dataset: ")
print(iris_df.isnull().sum())
```

```python
iris_df_ffill = iris_df.ffill()
iris_df_mean = iris_df.fillna(iris_df.mean())
iris_df_median = iris_df.fillna(iris_df.median())
iris_df_zero = iris_df.fillna(0)
# Expand iris_df with filled columns
iris_df_expanded = pd.concat([iris_df, iris_df_ffill.add_suffix('_ffill'), iris_df_mean.add_suffix('_mean'),iris_df_median.add_suffix('_median'),iris_df_zero.add_suffix('_zero')], axis=1)
# Display the head of the Expanded DataFrames
print("\nDataset after filling missing values:")
print(iris_df_expanded.head())
```

```python
# ==========Trimming Whitespace============
df = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})
df['Name'] = df['Name'].str.strip()

# ============Changing Datatype:=============
df = pd.DataFrame({'Age': ['25', '30', '35']})
# Change 'Age'  column data type to integer
df['Age'] = df['Age'].astype(int)
print(df)
# ===========Renaming columns:==============
# Rename columns
df = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})
df = df.rename(columns = {'Name': 'First Name', 'Age': 'Age in Years'})
print(df)
# ===========Removing Duplicates:=============
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice'], 'Age': [25, 30, 25]})
df = df.drop_duplicates()
print(df)
```

```python
import pandas as pd

medical_students_dataset = pd.read_csv("/content/medical_students_dataset.csv")
medical_students_df = pd.DataFrame(medical_students_dataset)
print(f"The DataFrame from medical_students.csv: \n{medical_students_df}")
```

```python
medical_students_dataset_null = medical_df.isnull().any()
print(f"Columns with Null value in medical_students_dataset.csv: \n{medical_students_dataset_null}")
```

```python
# medical_students_df = medical_students_df.dropna(subset=['Student ID'])

missing_students_id = medical_students_df['Student ID'].fillna(
    pd.Series(medical_students_df.index + 1, index = medical_students_df.index),
    )
medical_students_df['Student ID'] = missing_students_id

column_catagorical = ['Gender', 'Blood Type', 'Diabetes', 'Smoking']
for col in column_catagorical:
    medical_students_df[col] = medical_students_df[col].fillna(
        medical_students_df[col].mode()[0])

column_numerical = ['Height', 'Weight', 'Temperature', 'Heart Rate',
                 'Blood Pressure', 'Cholesterol']
for col in column_numerical:
    medical_students_df[col] = medical_students_df[col].fillna(
        medical_students_df[col].mean()
        )

missing_students_age = medical_students_df['Age'].fillna(
    medical_students_df['Age'].median()
    )
medical_students_df['Age'] = missing_students_age

medical_df['BMI'] = medical_df['Weight'] / (medical_df['Height']/100)**2
```

```python
dataset_duplicate_rows = medical_students_df.duplicated()
print(f"Number of duplicate rows: \n{dataset_duplicate_rows.sum()}")

medical_students_unique_df = medical_students_df.drop_duplicates()
print(f"New DataFrame after removing duplicate rows: \n{medical_students_unique_df}")
print(f"Number of duplicate rows: \n{medical_students_unique_df.duplicated().sum()}")
```

```python
import pandas as pd

titanic_dataset = pd.read_csv("/content/Titanic-Dataset.csv")
titanic_df = pd.DataFrame(titanic_dataset)

subset_columns = ['Name', 'Pclass', 'Sex', 'Age', 'Fare', 'Survived']
titanic_subset_df = titanic_df[subset_columns]
print(f"Subset DataFrame: \n{titanic_subset_df}")

first_class_passengers = titanic_df[titanic_subset_df['Pclass'] == 1]
print(f"First Class Passengers: \n{first_class_passengers}\n")

fare_mean = first_class_passengers['Fare'].mean()
fare_median = first_class_passengers['Fare'].median()
fare_maximum = first_class_passengers['Fare'].max()
fare_minimum = first_class_passengers['Fare'].min()

print(f"Fare statistics for first-class passengers: ")
print("=============================================")
print(f"Average cost for the Fare: {fare_mean}")
print(f"Median for the Fare: {fare_median}")
print(f"Maximum price for the Fare: {fare_maximum}")
print(f"Minimum price for the Fare: {fare_minimum}")
```

```python
missing_age_passenger = titanic_subset_df['Age'].isnull().sum()
print(f"Total number of missing values in Age column: {missing_age_passenger}")

updated_age_passenger = titanic_subset_df.dropna(subset = ['Age'])
print(f"Updated DataFrame after removing missing values: \n{updated_age_passenger}")
```

```python
titanic_df = titanic_df.dropna(subset=['Embarked'])

embarked_titanic = pd.get_dummies(titanic_df['Embarked'], prefix='Embarked')

titanic_df = pd.concat([titanic_df, embarked_titanic], axis = 1)

original_embarked_droped_df = titanic_df.drop('Embarked', axis=1)

print(original_embarked_droped_df.head())
```
