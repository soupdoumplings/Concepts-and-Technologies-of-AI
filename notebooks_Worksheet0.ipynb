```python
#TASK TO-DO
time_data = [
    (3.5, 2.0, 7.0), (5.0, 1.5, 6.5), (2.5, 3.0, 8.0),
    (4.0, 2.0, 6.0), (1.5, 4.5, 9.0), (3.0, 2.5, 7.5),
    (5.5, 1.0, 6.0), (2.0, 3.5, 8.5), (4.5, 2.0, 7.0),
    (3.0, 3.0, 7.5), (6.0, 1.5, 6.0), (2.5, 4.0, 8.0),
    (4.0, 2.5, 7.0), (5.0, 2.0, 6.5), (3.5, 2.5, 7.0)
]

#TASK 1
#1 CREATE EMPTY LIST FOR STUDY TIME CLASSIFICATION LOW, MODERATE, HIGH

Low = []
Moderate = []
High = []

#2 ITERATE OVER THE time_data LIST AND ADD EACH STUDY HOUR TO THE APPROPRIATE CATEGORY.

for time in time_data:
    if time[0]>=3 and time[0]<=5:
        Moderate.append(time[0])
        
    elif time[0] < 3:
        Low.append(time[0])            
    else:
        High.append(time[0])
    
#3 PRINT THE LIST TO VERIFY

print(f"Updated Low List {Low}")
print(f"Updated Moderate List {Moderate}")
print(f"Updated High List {High}")
```

```python
def sum_of_nested_lists(nested_list):
    total = 0
    for list_element in nested_list:
        if isinstance(list_element, list):
            total += sum_of_nested_lists(list_element)
        else:
            total += list_element
    return total

print(sum_of_nested_lists([1, [2, [3, 4], 5], 6, [7, 8]]))
        
```

```python
def generatative_permutation(stg):
    if len(stg) <= 1:
        return(stg)

    permutation = set()

    for ch in range(len(stg)):

        char = stg[ch]
        remaining = stg[:ch] + stg[ch+1:]

        for p in generatative_permutation(remaining):
            permutation.add(char + p)

    return list(permutation)

print(generatative_permutation("abc"))
    
            
```

```python
def calculation_of_directory_size(directory):
    total = 0
    for key, value in directory.items():
        if isinstance(value, dict):
            total += calculation_of_directory_size(value)
        else:
            total += value
    return total
directory_structure = {
"file1.txt": 200,
"file2.txt": 300,
"subdir1": {
"file3.txt": 400,
"file4.txt": 100
},
"subdir2": {
"subsubdir1": {
"file5.txt": 250
},
"file6.txt": 150
}
}
print(calculation_of_directory_size(directory_structure))
```

```python
def min_coins(coins, amount):
    dp = [float('inf')] * (amount + 1)

    dp[0] = 0

    for coin in coins:
        for z in range(coin, amount + 1):
            dp[z] = min(dp[z], dp[z-coin]+1)

    return dp[amount] if dp[amount] != [float('inf')] else -1

coins = [1, 2, 5]
amount = 11
print(min_coins(coins, amount))
```

```python
def longest_common_subsequence(s1, s2):
    m, n = len(s1), len(s2)

    dp = [[0] * (n + 1) for _ in range (m +1)]

    for i in range(1, m+1):
        for j in range(1, n+1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp [i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp [i - 1][j], dp [i][j - 1])

    return dp[m][n]

s1 = "abcde"
s2 = "ace"

print(longest_common_subsequence(s1, s2))

    
```

```python
def knapsack(weights, values, capacity):
    n = len(weights)

    dp = [[0] * (capacity + 1) for _ in range (n + 1)]

    for i in range(1, n+1):
        w = weights[i - 1]
        v = values[i - 1]

        for c in range(1, capacity + 1):
            if w <= c:
                dp[i][c] = max(dp[i - 1][c], dp[i - 1][c - w] +v)
            else:
                dp[i][c] = dp[i - 1][c]
    return dp[n][capacity]

weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7
print(knapsack(weights, values, capacity))      
```
