Problem - 4: Matrix Operations:



*  For the following arrays:

    A = np.array([[3,4],[7,8]]) and B = np.array([[5,3],[2,1]]);
    
    Prove following with Numpy:
    1. Prove A.A^−1 = I.
    2. Prove AB̸ = BA.
    3. Prove (AB)^T = B^T A^T
*   Solve the following system of Linear equation using Inverse Methods.

    2x − 3y + z = −1

    x − y + 2z = −3
    
    3x + y − z = 9
  

*   Now: solve the above equation using np.linalg.inv function.





```python
# Given arrays:

A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])
```

```python
# 1. Prove A.A^−1 = I.

# L.H.S
A_inverse = np.linalg.inv(A)
A_dot_Ainv = np.dot(A, A_inverse)
print(f"L.H.S: \n{A_dot_Ainv}")

# RHS
I = np.identity(2)
print(f"R.H.S: \n{I}")

# verify
verify = np.allclose(A_dot_Ainv, I)
print(f"Hence, A.A^−1 = I is {verify}")
```

```python
# 2. Prove AB̸= BA.

# L.H.S
A_dot_B = np.dot(A, B)
print(f"L.H.S: \n{A_dot_B}")

# R.H.S
B_dot_A = np.dot(B, A)
print(f"R.H.S: \n{B_dot_A}")

# verify
verify = np.allclose(A_dot_B, B_dot_A)
print(f"Hence, AB̸= BA is {not verify}")
```

```python
# 3. Prove (A.B)^T = B^T.A^T

# L.H.S
A_dot_B = np.dot(A, B)
AB_trans = np.transpose(A_dot_B)
print(f"L.H.S: \n{AB_trans}")

# R.H.S
B_trans = np.transpose(B)
A_trans = np.transpose(A)
Btrans_dot_Atrans = np.dot(B_trans, A_trans)
print(f"R.H.S: \n{Btrans_dot_Atrans}")

# verify
verify = np.allclose(AB_trans, Btrans_dot_Atrans)
print(f"Hence, (A.B)^T = B^T.A^T is { verify}")
```

```python
# Solve the following system of Linear equation using Inverse Methods.

# 2x − 3y + z = −1

# x − y + 2z = −3

# 3x + y − z = 9

# Initialize the given linear equations into matrix equation form AX = B

A = np.array([[2, 3, 1], [1, -1, 2], [3, 1, -1]])
X = np.array(["x", "y", "z"])
B = np.array([-1, -3, 9])


```

```python
# Now: solve the above equation using np.linalg.inv function.

# Equation: AX = B; X = A^-1.B

A_inverse = np.linalg.inv(A)
X = np.dot(A_inverse, B)

print(f"The value of X: {X[0]}; Y: {X[1]}; Z: {X[2]}")
```

Experiment: How Fast is Numpy?

```python
# 1. Element-wise Addition:

import random

from timeit import timeit

def add_python(r1, r2):
 return [r1 + r2 for r1, r2 in zip(r1, r2)]

def add_numpy(r1, r2):
 return np.add(r1, r2)

list1 = random.sample(range(1000000), 1000000)
list2 = random.sample(range(1000000), 1000000)

array1 = np.array(list1)
array2 = np.array(list2)

# Using Python list:
time_python = timeit("add_python(list1, list2)", globals = globals(), number = 1000)
print(f"Time taken by python list to execute the operation: {time_python}")

# Using Numpy array:
time_numpy = timeit("add_numpy(array2, array2)", globals = globals(), number = 1000)
print(f"Time taken by numpy array to execute the operation: {time_numpy}")
```

```python
# 2. Element-wise Multiplication:

import random

from timeit import timeit

def multiply_python(r1, r2):
 return [r1 * r2 for r1, r2 in zip(r1, r2)]

def multiply_numpy(r1, r2):
 return np.multiply(r1, r2)

list1 = random.sample(range(1000000), 1000000)
list2 = random.sample(range(1000000), 1000000)

array1 = np.array(list1)
array2 = np.array(list2)

# Using Python list:
time_python = timeit("multiply_python(list1, list2)", globals = globals(), number = 1000)
print(f"Time taken by python list to execute the operation: {time_python}")

# Using Numpy array:
time_numpy = timeit("multiply_numpy(array1, array2)", globals = globals(), number = 1000)
print(f"Time taken by numpy array to execute the operation: {time_numpy}")
```

```python
# 3. Element-wise dot product:

import random

from timeit import timeit

def dot_python(r1, r2):
 return sum(r1 * r2 for r1, r2 in zip(r1, r2))

def dot_numpy(r1, r2):
 return np.dot(r1, r2)

list1 = random.sample(range(1000000), 1000000)
list2 = random.sample(range(1000000), 1000000)

array1 = np.array(list1)
array2 = np.array(list2)

# Using Python list:
time_python = timeit("dot_python(list1, list2)", globals = globals(), number = 1000)
print(f"Time taken by python list to execute the operation: {time_python}")

# Using Numpy array:
time_numpy = timeit("dot_numpy(array1, array2)", globals = globals(), number = 1000)
print(f"Time taken by numpy array to execute the operation: {time_numpy}")
```
