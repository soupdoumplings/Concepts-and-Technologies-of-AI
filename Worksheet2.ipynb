Handling Missing Values: Example: 12 - Adding Missing Values

````python
import pandas as pd
from sklearn.datasets import load_iris
import numpy as np
iris = load_iris()
iris_df = pd.DataFrame(data = np.c_[iris['data'], iris['target']], columns = iris['feature_names'] + ['target'])
np.random.seed(42)
mask = np.random.rand(*iris_df.shape) < 0.1
iris_df[mask] = np.nan
print("Missing values in Iris Dataset: ")
print(iris_df.isnull().sum())
````

Handling Missing Values: Example: 13 - Techniques for Filling Missing Values.

````python
iris_df_ffill = iris_df.ffill()
iris_df_mean = iris_df.fillna(iris_df.mean())
iris_df_median = iris_df.fillna(iris_df.median())
iris_df_zero = iris_df.fillna(0)
# Expand iris_df with filled columns
iris_df_expanded = pd.concat([iris_df, iris_df_ffill.add_suffix('_ffill'), iris_df_mean.add_suffix('_mean'),iris_df_median.add_suffix('_median'),iris_df_zero.add_suffix('_zero')], axis=1)
# Display the head of the Expanded DataFrames
print("\nDataset after filling missing values:")
print(iris_df_expanded.head())
````

Example: 15 - Data Transformation: Reshaping DataFrame using Pivot and Melt.

````python
import pandas as pd
# Sample DataFrame
data = {
    'Date': [
        '2024-01-01', '2024-01-01', '2024-01-01', '2024-01-01',
        '2024-01-02', '2024-01-02', '2024-01-02', '2024-01-02',
        '2024-01-03', '2024-01-03', '2024-01-03', '2024-01-03'
    ],
    'City': [
        'Kathmandu', 'Pokhara', 'Biratnagar', 'Lalitpur',
        'Kathmandu', 'Pokhara', 'Biratnagar', 'Lalitpur',
        'Kathmandu', 'Pokhara', 'Biratnagar', 'Lalitpur'
    ],
    'Temperature (°C)': [
        15, 18, 20, 16,
        16, 19, 21, 17,
        14, 17, 19, 15
    ]
}

df = pd.DataFrame(data)
# pivot: Reshape Data to show cities as columns
pivoted_df = df.pivot(index='Date', columns='City', values='Temperature (°C)')
print(pivoted_df)

# ================Melting:=================
melted_df = pd.melt(pivoted_df.reset_index(), id_vars=['Date'], var_name='City', value_name='Temperature (°C)')
print(melted_df)

````

Example: 20 - Merging DataFrames using merge

````python
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Names': ['Jonathan', 'Dio', 'Valentine']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Stands': ['The World', 'Dirty Deeds Done Dirt Cheap', 'Stardust Platinum']})

inner_merged = pd.merge(df1, df2, on = 'ID', how = 'inner')
print("Inner Merge:")
print(inner_merged)

left_merged = pd.merge(df1, df2, on = 'ID', how = 'left')
print("Left Merge:")
print(left_merged)

outer_merged = pd.merge(df1, df2, on = 'ID', how = 'outer')
print("Outer Merge:")
print(outer_merged)
````

###**Question: 3 - Fill the missing values using various techniques.**

````python
# medical_students_df = medical_students_df.dropna(subset=['Student ID'])

missing_students_id = medical_students_df['Student ID'].fillna(
    pd.Series(medical_students_df.index + 1, index = medical_students_df.index),
    )
medical_students_df['Student ID'] = missing_students_id

"""
Since the Values of Student ID must be Unique, It cannont be filled with any
other methods and must be filled with a manual calculation. Since the DataFrame
contains a serial Index, adding it by 1 gives us the "Student ID" for the
rows with missing value.

Observation: pd.Series(); Creates a One-dimensional labeled data structure,
                          with new "Student ID" as its value and the index of
                          the missing value as its index.
             fillna(); Fills the missing values in DataFrame.
"""

column_catagorical = ['Gender', 'Blood Type', 'Diabetes', 'Smoking']
for col in column_catagorical:
    medical_students_df[col] = medical_students_df[col].fillna(
        medical_students_df[col].mode()[0])

"""
The Values in the column of column_catagorical are non numerical values,
performing calulation methods i.e. mean(), median() is illogical. Hence, mode()
method is used to fill the missing values.

Observation: for col in column_catagorical: loops through every column in the
             column_catagorical list.
             mode(); Returns the mode of the values in the column.
             fillna(); Fills the missing values in DataFrame.
"""

column_numerical = ['Height', 'Weight', 'Temperature', 'Heart Rate',
                 'Blood Pressure', 'Cholesterol']
for col in column_numerical:
    medical_students_df[col] = medical_students_df[col].fillna(
        medical_students_df[col].mean()
        )
"""
The Values in the column of column_numerical are numerical values,
so calulation methods i.e. mean(), median() method can used to fill the missing
values. Furthermore, since the values from columns in column_numerical is all
mathematical scale based i.e. Height = cm/m and relatively normally distributed,
using mean() method is more appropriate.

Median() could also be used if there are outliers in the values.

Observation: for col in column_numerical: loops through every column in the
             column_numerical list.
             mean(); Returns the mean of the values in the column.
             fillna(); Fills the missing values in DataFrame.
"""

missing_students_age = medical_students_df['Age'].fillna(
    medical_students_df['Age'].median()
    )
medical_students_df['Age'] = missing_students_age

"""
The missing values are filled using median() method because it is less affected
by outliers and age is not a mathematical scale. Making median() more appropriate
over mean() method.

Observation: median(); Returns the median of the values in the column.
             fillna(); Fills the missing values in DataFrame.
"""

medical_students_df['BMI'] = medical_students_df['Weight'] / (medical_students_df['Height']/100)**2

"""
Since before this step both the Height and Weight column is filled with no
missing values. The BMI column can be directly filled with the formula.
"""
````
